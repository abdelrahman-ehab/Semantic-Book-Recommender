```python
#mapping the broad categories into a smaller non broad to help in filtering
# for the top 12 categories available
category_mapping = {'Fiction' : "Fiction",
'Juvenile Fiction': "Children's Fiction",
'Biography & Autobiography': "Nonfiction",
'History': "Nonfiction",
'Literary Criticism': "Nonfiction",
'Philosophy': "Nonfiction",
'Religion': "Nonfiction",
'Comics & Graphic Novels': "Fiction",
'Drama': "Fiction",
'Juvenile Nonfiction': "Children's Nonfiction",
'Science': "Nonfiction",
'Poetry': "Fiction"}

books["simple_categories"] = books["categories"].map(category_mapping)
```

**Zero-Shot Classification**

```python
from transformers import pipeline
fiction_categories = ["Fiction", "Nonfiction"]
pipe = pipeline("zero-shot-classification",model = "facebook/bart-large-mnli",device= "cuda")
```

```python
books.loc[books["simple_categories"]=="Fiction","description"].reset_index(drop=True)[0]
```

```python
sequence = books.loc[books["simple_categories"]=="Fiction","description"].reset_index(drop=True)[0]

pipe(sequence,fiction_categories)
```

```python
import numpy as np
# Taking the maximum probability provided by the classifier
max_index = np.argmax(pipe(sequence,fiction_categories)["scores"])
# Getting the label which corresponds to the max prob
max_label= pipe(sequence,fiction_categories)["labels"][max_index]
max_label
```

```python
def generate_classifications(sequence, categories):
    predictions = pipe(sequence, categories)
    max_index = np.argmax(pipe(sequence,categories)["scores"])
    max_label= pipe(sequence,categories)["labels"][max_index]
    return max_label
```

Measuring The classifier Accuracy

```python
from tqdm import tqdm

actual_cats = []
predicted_cats = []

for i in tqdm(range(0,300)):
  sequence = books.loc[books["simple_categories"]== "Fiction","description"].reset_index(drop=True)[i]
  actual_cats.append("Fiction")
  predicted_cats.append(generate_classifications(sequence, fiction_categories))

```

```python
for i in tqdm(range(0,300)):
  sequence = books.loc[books["simple_categories"]== "Nonfiction","description"].reset_index(drop=True)[i]
  actual_cats.append("Nonfiction")
  predicted_cats.append(generate_classifications(sequence, fiction_categories))
```

```python
predicted_df = pd.DataFrame({"actual": actual_cats, "predicted": predicted_cats})
predicted_df
```

```python
predicted_df["correct_prediction"] = (
    np.where(predicted_df["actual"] == predicted_df["predicted"], 1, 0)
)
```

```python
predicted_df["correct_prediction"].sum() / len(predicted_df)
```

```python
isbns = []
predicted_cats = []

missing_cats = books.loc[books["simple_categories"].isna(), ["isbn13", "description"]].reset_index(drop=True)
```

```python
for i in tqdm(range(0, len(missing_cats))):
    sequence = missing_cats["description"][i]
    predicted_cats += [generate_classifications(sequence, fiction_categories)]
    isbns += [missing_cats["isbn13"][i]]
```

```python
missing_predicted_df = pd.DataFrame({"isbn13": isbns, "predicted_categories": predicted_cats})
```

```python
missing_predicted_df
```

```python
books = pd.merge(books, missing_predicted_df, on="isbn13", how="left")
books["simple_categories"] = np.where(books["simple_categories"].isna(), books["predicted_categories"], books["simple_categories"])
books = books.drop(columns = ["predicted_categories"])
```
